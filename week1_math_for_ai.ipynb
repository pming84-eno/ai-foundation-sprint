```python
import numpy as np

def dot(a, b):
    return np.dot(a, b)

def cosine_sim(a, b):
    return dot(a, b) / (np.linalg.norm(a) * np.linalg.norm(b))

def euclidean(a, b):
    return np.linalg.norm(a - b)
```

```python
# synthetic data
np.random.seed(0)
X = np.random.randn(100, 1)
y = 3 * X.squeeze() + 5 + np.random.randn(100) * 0.5

# init params
w, b = 0.0, 0.0
lr = 0.01
n_iter = 1000

for _ in range(n_iter):
    y_pred = X * w + b
    dw = (2/len(X)) * X.T @ (y_pred - y)
    db = (2/len(X)) * np.sum(y_pred - y)
    w -= lr * dw
    b -= lr * db
```
